```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Use this in Jupyter for inline animations
from IPython.display import HTML

# Set random seed
np.random.seed(11)

# Original non-normal data
original_data = np.random.exponential(scale=2.0, size=10000)

# Sample sizes to demonstrate
sample_sizes = [2, 5, 10, 20, 30, 50, 75, 100, 200]
num_samples = 500

# Set up figure and axes
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

# Initialization function
def init():
    ax[0].clear()
    ax[0].hist(original_data, bins=50, density=True, color='skyblue', edgecolor='black')
    ax[0].set_title('Original Distribution (Exponential)')
    ax[0].set_xlabel('Value')
    ax[0].set_ylabel('Density')


# Update function for animation
def update(frame):
    n = sample_sizes[frame]
    sample_means = [np.mean(np.random.choice(original_data, size=n, replace=True)) for _ in range(num_samples)]

    ax[1].clear()
    ax[1].hist(sample_means, bins=50, density=True, color='salmon', edgecolor='black')
    ax[1].set_title(f'Distribution of Sample Means (n={n})')
    ax[1].set_xlabel('Sample Mean')
    ax[1].set_ylabel('Density')
    ax[1].set_xlim(0, 10)  # Adjust based on expected mean range

    fig.suptitle(f'Central Limit Theorem Animation - Sample Size = {n}', fontsize=16)

# Create animation
ani = FuncAnimation(fig, update, frames=len(sample_sizes), init_func=init,
                    repeat=False, interval=1500)

# Display animation in notebook
HTML(ani.to_jshtml())
```
