```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Set seed for reproducibility
np.random.seed(42)

# Define parameters for the two normal distributions
mu1, sigma1, n1 = 5, 1.0, 500
mu2, sigma2, n2 = 10, 0.5, 500

# Generate samples
sample1 = np.random.normal(mu1, sigma1, n1)
sample2 = np.random.normal(mu2, sigma2, n2)
combined_sample = np.concatenate([sample1, sample2])

# Estimate parameters for combined distribution
combined_mu = np.mean(combined_sample)
combined_sigma = np.std(combined_sample)

# Prepare x-axis for plotting PDFs
x = np.linspace(min(combined_sample) - 2, max(combined_sample) + 2, 1000)

# Plot histograms and theoretical PDFs
plt.figure(figsize=(10, 6))

# Plot histograms (normalized)
plt.hist(sample1, bins=30, density=True, alpha=0.4, label='Sample 1 (N1)', color='blue')
plt.hist(sample2, bins=30, density=True, alpha=0.4, label='Sample 2 (N2)', color='orange')
plt.hist(combined_sample, bins=30, density=True, alpha=0.4, label='Combined Sample', color='green')

# Plot theoretical PDFs
plt.plot(x, norm.pdf(x, mu1, sigma1), 'b--', label='N(μ1, σ1²)')
plt.plot(x, norm.pdf(x, mu2, sigma2), 'orange', linestyle='--', label='N(μ2, σ2²)')
plt.plot(x, norm.pdf(x, combined_mu, combined_sigma), 'g-', label='Combined N(μ, σ²)')

# Final touches
plt.title('Approximation of Two Normals by a Single Normal')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
